This template is used to read one or more .diag files using nrgpy. Each diagnostic signal across all files is plotted.

```python
import pandas as pd

from se_analysis import project, read_files
from se_analysis.plot import group_signals, plot_signals
```

Set up data directory by using `project` to find your configured project directory. This is set in the `SE_PROJECTS_DIR` variable in the .env file.

```python
data_dir = project("SRM", "Sundance_CO", "data")
```

Now read the diagnostic files in that directory.

```python
reader = read_files(data_dir, file_type="diag")
```

Display the site info to make sure you're using the correct data files.

```python
reader.Site_info
```

Copy `reader.data` to `df`, which is more convenient to use. Create the timestamp index and inspect the result.

```python
df = reader.data
df["Stats_Timestamp"] = pd.to_datetime(df["Stats_Timestamp"])
df = df.set_index("Stats_Timestamp").sort_index()
df
```

Inspect the columns of the DataFrame.

```python
for col in df.columns:
    print(col)
```

This function groups the columns into their statistics so that plots show related columns together.

```python
groups = group_signals(df)
groups
```

Plot every signal to visually inspect. Statistics of the same signal share an axes except SD is shown separately since it usually has a much smaller scale.

```python
plot_signals(df)
```
